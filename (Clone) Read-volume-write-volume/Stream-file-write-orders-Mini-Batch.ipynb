```python
df_orders = (
  spark.read.format("csv")
  .option("header", "true")
  .load("/Volumes/project_kankan/orders/orders_volume/data/")
)

df_orders.write.format("delta") \
  .option("checkpointLocation", "/Volumes/project_kankan/orders/orders_volume/checkpoint") \
  .mode("overwrite") \
  .save("/Volumes/project_kankan/orders/orders_volume_tgt/delta")
```
