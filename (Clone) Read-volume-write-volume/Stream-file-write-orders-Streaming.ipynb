```python
df_orders_stream = (
  spark.readStream
  .format("cloudFiles")
  .option("cloudFiles.format", "csv")
  .option("header", "true")
  .option("inferSchema", "true")
  .option("cloudFiles.schemaEvolutionMode", "rescue")
  .option("cloudFiles.schemaLocation", "/Volumes/project_kankan/orders/orders_volume/schema")
  .option("maxFilesize", 1)
  .load("/Volumes/project_kankan/orders/orders_volume/data")
)

(
  df_orders_stream.writeStream
  .format("delta")
  .option("mergeSchema", "true")
  .option("checkpointLocation", "/Volumes/project_kankan/orders/orders_volume/checkpoint")
  .outputMode("append")
  .trigger(once=True)
  .start("/Volumes/project_kankan/orders/orders_volume_tgt/delta")
)
```
